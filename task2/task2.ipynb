```python
df = pd.read_csv("googleplaystore.csv")

df["Installs"] = (
    df["Installs"]
      .str.replace(",", "", regex=False)
      .str.replace("+", "", regex=False)
)

df["Installs"] = pd.to_numeric(df["Installs"], errors="coerce")

df = df.dropna(subset=["Category", "Installs"])
```

```python
exclude = ("A", "C", "G", "S")

filtered = df[
    ~df["Category"].str.upper().str.startswith(exclude)
]
```

```python
top5 = (
    filtered.groupby("Category", as_index=False)["Installs"]
            .sum()
            .sort_values("Installs", ascending=False)
            .head(5)
)
```

```python
top5["Highlight"] = top5["Installs"].apply(
    lambda x: "Above 1 Million" if x > 1_000_000 else "Below 1 Million"
)
```

```python
country_map = {
    "BUSINESS": "India",
    "BEAUTY": "France",
    "DATING": "United States",
    "EVENTS": "Germany",
    "FAMILY": "Brazil",
    "FINANCE": "Canada",
    "FOOD_AND_DRINK": "Italy",
    "HEALTH_AND_FITNESS": "Australia",
    "LIBRARIES_AND_DEMO": "United Kingdom",
    "MAPS_AND_NAVIGATION": "Japan",
    "MEDICAL": "South Africa",
    "TOOLS": "Singapore",
    "PRODUCTIVITY": "Netherlands",
    "PERSONALIZATION": "Spain"
}

top5["Country"] = top5["Category"].map(country_map)
top5 = top5.dropna(subset=["Country"])
```

```python
fig = px.choropleth(
    top5,
    locations="Country",
    locationmode="country names",
    color="Installs",
    hover_name="Category",
    hover_data={
        "Highlight": True,
        "Installs": ":,.0f"
    },
    color_continuous_scale="Viridis",
    title="Global Installs by Top 5 App Categories"
)

fig.update_layout(
    title_x=0.5,
    geo=dict(showframe=False, showcoastlines=True)
)
```

```python
from datetime import datetime
import pytz

india = pytz.timezone("Asia/Kolkata")
current_time = datetime.now(india)

if 18 <= current_time.hour < 20:
    fig.show()
else:
    print("This graph is available only between 6 PM and 8 PM IST.")
```
